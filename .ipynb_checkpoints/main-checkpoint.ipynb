```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools
from math import sqrt

# helper for nicer DataFrame display in this environment (the tool will handle it)
try:
    from caas_jupyter_tools import display_dataframe_to_user
except Exception:
    display_dataframe_to_user = None

n = 1000          # number of points
A = 1.0           # end coordinate on each axis (start is 0)
k = 10            # parameter (should be in [4, 40])
interpretation = 'variance'  # 'variance' or 'std' -- here we assume the given numbers are variances
random_seed = 42

np.random.seed(random_seed)

# compute the per-axis dispersion values according to the problem statement
# given numbers: axis1 = A/k, axis2 = A/k, axis3 = A/(k*2), axis4 = A/(k*4), axis5 = A/(k*8)
raw_vals = np.array([A/k, A/k, A/(k*2), A/(k*4), A/(k*8)], dtype=float)

if interpretation == 'variance':
    variances = raw_vals.copy()
    sds = np.sqrt(variances)
elif interpretation == 'std':
    sds = raw_vals.copy()
    variances = sds**2
else:
    raise ValueError("interpretation must be 'variance' or 'std'")

# create param t along the line from 0 to A
# use evenly spaced points so the line progression is visible; you can change to random if desired
t = np.linspace(0, A, n)  # shape (n,)

# create the noiseless points along the diagonal (each point = [t_i]*5)
base = np.tile(t.reshape(-1,1), (1,5))  # shape (n,5)

# noise: independent Gaussian per axis with given variance
noise = np.random.normal(loc=0.0, scale=sds.reshape(1,-1), size=(n,5))

# final dataset
X = base + noise

# put into DataFrame
cols = [f'x{i+1}' for i in range(5)]
df = pd.DataFrame(X, columns=cols)

# show dataset head using the environment helper if available
if display_dataframe_to_user is not None:
    display_dataframe_to_user("Generated 5D dataset (first rows)", df.head(10))
else:
    print("First 10 rows of generated dataset:")
    display(df.head(10))

# report theoretical and empirical statistics
print("\nParameters:")
print(f"  n = {n}")
print(f"  A = {A}")
print(f"  k = {k}  (should be in [4,40])")
print(f"  interpretation = '{interpretation}'\n")

for i, (var, sd) in enumerate(zip(variances, sds)):
    print(f"axis {i+1}: variance = {var:.6g}, std = {sd:.6g}")

empirical_vars = df.var(axis=0, ddof=0)  # population variance (ddof=0)
print("\nEmpirical variances (population, ddof=0):")
for i, v in enumerate(empirical_vars):
    print(f"  {cols[i]}: {v:.6g}")

# Optionally compute PCA to show that most variance is along the first principal component
from numpy.linalg import eig
X_centered = X - X.mean(axis=0)
cov = np.cov(X_centered, rowvar=False, ddof=0)
eigs, eigvecs = eig(cov)
# sort eigenvalues descending
idx = np.argsort(eigs)[::-1]
eigs_sorted = eigs[idx].real
explained = eigs_sorted / eigs_sorted.sum()
print("\nTop 5 eigenvalues of covariance matrix and explained variance ratio:")
for i, (ev, ex) in enumerate(zip(eigs_sorted, explained)):
    print(f"  PC{i+1}: eigenvalue={ev:.6g}, explained={ex:.3%}")

# -------------------------
# Pairwise scatter plots (each in its own figure)
# -------------------------
pairs = list(itertools.combinations(range(5), 2))
print(f"\nGenerating {len(pairs)} pairwise scatter plots (each opens separately).")

for (i, j) in pairs:
    plt.figure(figsize=(4.5,4.5))
    plt.scatter(df[cols[i]], df[cols[j]], s=12, alpha=0.6)
    plt.xlabel(cols[i])
    plt.ylabel(cols[j])
    plt.title(f"{cols[i]} vs {cols[j]}  (var{i+1}={variances[i]:.3g}, var{j+1}={variances[j]:.3g})")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

csv_path = "./generated_5d_dataset.csv"
df.to_csv(csv_path, index=False)
print(f"\nDataset saved to: {csv_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import os

# Загрузка датасетов df_out_i
datasets = {}
for i in [1, 2, 5, 10]:
    path = f"./df_out_{i}.csv"
    if os.path.exists(path):
        datasets[f"df_out_{i}"] = pd.read_csv(path)

cols = [f'x{k}' for k in range(1,6)]
pc_pairs = [(i,j) for i in range(5) for j in range(i+1,5)]

for name, df_out in datasets.items():
    X = df_out.values
    scaler = StandardScaler()
    X_std = scaler.fit_transform(X)
    
    pca = PCA()
    X_pca = pca.fit_transform(X_std)
    
    print(f"\n=== {name} (n={len(df_out)}) ===")
    
    # 1. Scree plot
    plt.figure(figsize=(5,3.5))
    plt.bar(range(1,6), pca.explained_variance_ratio_, alpha=0.7, color='steelblue')
    plt.plot(range(1,6), pca.explained_variance_ratio_, 'o-', color='red')
    plt.title(f"Scree plot — {name}")
    plt.xlabel("PC"); plt.ylabel("Explained variance ratio")
    plt.grid(True); plt.show()
    
    # Корреляции признаков с PC (для круstes круг корреляций)
    corr = np.corrcoef(X_std.T, X_pca.T)[:5, 5:]
    
    # Единый масштаб для векторов переменных (loadings)
    # Выбираем масштаб так, чтобы стрелки занимали ~30–40% от диапазона точек
    # Единый фиксированный масштаб стрелок для всех пар PC
    data_range = np.ptp(X_pca, axis=0).max()           # исправлено для NumPy 2.0+
    desired_arrow_len = data_range * 0.35              # стрелки ~35% от размаха точек
    max_loading = np.abs(pca.components_).max()
    scale_factor = desired_arrow_len / max(1.0, max_loading)   # один масштаб на весь датасет
    
    for i, j in pc_pairs:
        fig, axs = plt.subplots(1, 3, figsize=(15, 5))
        fig.suptitle(f"{name}: PC{i+1} vs PC{j+1}", fontsize=14)
        
        # 2. Только точки
        axs[0].scatter(X_pca[:,i], X_pca[:,j], s=15, alpha=0.7, c='steelblue')
        axs[0].set_title("Точки")
        axs[0].grid(True)
        
        # 3. Точки + векторы переменных (одинаковый масштаб!)
        axs[1].scatter(X_pca[:,i], X_pca[:,j], s=15, alpha=0.7, c='steelblue')
        for k in range(5):
            xi = pca.components_[i, k] * scale_factor
            yj = pca.components_[j, k] * scale_factor
            axs[1].arrow(0, 0, xi, yj, color='red', width=0.02,
                        head_width=0.15, head_length=0.15)
            axs[1].text(xi*1.15, yj*1.15, cols[k], color='red',
                       ha='center', fontsize=11)
        axs[1].set_title("Точки + векторы переменных")
        axs[1].grid(True)
        
        # 4. Круг корреляций (всегда [-1,1])
        circle = plt.Circle((0,0), 1, color='gray', fill=False, ls='--')
        axs[2].add_artist(circle)
        for k in range(5):
            axs[2].arrow(0, 0, corr[k,i], corr[k,j], color='red', width=0.01,
                        head_width=0.08, head_length=0.08)
            axs[2].text(corr[k,i]*1.1, corr[k,j]*1.1, cols[k],
                       color='red', ha='center', fontsize=11)
        axs[2].set_xlim(-1.2,1.2); axs[2].set_ylim(-1.2,1.2)
        axs[2].set_aspect('equal')
        axs[2].set_title("Круг корреляций")
        axs[2].grid(True)
        axs[2].axhline(0,color='k',lw=0.8); axs[2].axvline(0,color='k',lw=0.8)
        
        for ax in axs:
            ax.set_xlabel(f"PC{i+1}")
            ax.set_ylabel(f"PC{j+1}")
        
        plt.tight_layout()
        plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import umap
import pacmap
from sklearn.manifold import TSNE
import os

datasets = {'df': df}
for i in [1, 2, 5, 10]:
    path = f"./df_out_{i}.csv"
    if os.path.exists(path):
        datasets[f"df_out_{i}"] = pd.read_csv(path)

methods = ['UMAP', 'PaCMAP', 't-SNE']
n_random = 4
random_states = [0, 42, 123, 999]

scaler = StandardScaler()

for name, data in datasets.items():
    X = data.values
    X_std = scaler.fit_transform(X)
    
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_std)
    
    print(f"\n=== Проекции для {name} (n={len(data)}) ===")
    
    embeddings = {}
    
    for rs in random_states:
        embeddings[f'UMAP_rs{rs}'] = umap.UMAP(n_components=2, random_state=rs).fit_transform(X_std)
        embeddings[f'PaCMAP_rs{rs}'] = pacmap.PaCMAP(n_components=2, random_state=rs).fit_transform(X_std)
        embeddings[f'tSNE_rs{rs}'] = TSNE(n_components=2, random_state=rs, perplexity=min(30, len(X)//4)).fit_transform(X_std)
    
    embeddings['PCA'] = X_pca
    
    fig, axes = plt.subplots(5, 3, figsize=(12, 18))
    fig.suptitle(f"Проекции: {name}", fontsize=16)
    
    row = 0
    for rs in random_states:
        axes[row, 0].scatter(embeddings[f'UMAP_rs{rs}'][:,0], embeddings[f'UMAP_rs{rs}'][:,1], s=10, alpha=0.7)
        axes[row, 0].set_title(f"UMAP (rs={rs})")
        
        axes[row, 1].scatter(embeddings[f'PaCMAP_rs{rs}'][:,0], embeddings[f'PaCMAP_rs{rs}'][:,1], s=10, alpha=0.7)
        axes[row, 1].set_title(f"PaCMAP (rs={rs})")
        
        axes[row, 2].scatter(embeddings[f'tSNE_rs{rs}'][:,0], embeddings[f'tSNE_rs{rs}'][:,1], s=10, alpha=0.7)
        axes[row, 2].set_title(f"t-SNE (rs={rs})")
        
        for c in range(3):
            axes[row, c].grid(True)
        row += 1
    
    # PCA в последней строке
    axes[4, 0].scatter(X_pca[:,0], X_pca[:,1], s=10, alpha=0.7, c='steelblue')
    axes[4, 0].set_title("PCA (2D)")
    axes[4, 0].grid(True)
    axes[4, 1].axis('off')
    axes[4, 2].axis('off')
    
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools

n_per_class = 1000
A = 1.0
sigma = A / 200000          # очень маленький шум
np.random.seed(42)

# Класс 0: линия от (0,0,0,0) → (A,A,A,A)
t0 = np.linspace(0, A, n_per_class)
base0 = np.tile(t0.reshape(-1, 1), (1, 4))
noise0 = np.random.normal(0, sigma, size=(n_per_class, 4))
class0 = base0 + noise0

# Класс 1: линия от (A/100,0,0,0) → (A+A/100,A,A,A)
t1 = np.linspace(A/100, A + A/100, n_per_class)
base1 = np.tile(t1.reshape(-1, 1), (1, 4))
noise1 = np.random.normal(0, sigma, size=(n_per_class, 4))
class1 = base1 + noise1

# Объединение
X = np.vstack([class0, class1])
y = np.hstack([np.zeros(n_per_class, dtype=int), 
               np.ones(n_per_class, dtype=int)])

# DataFrame
cols = [f'x{i+1}' for i in range(4)]
df_LDA = pd.DataFrame(X, columns=cols)
df_LDA['class'] = y

df_LDA.to_csv("./df_LDA.csv", index=False)
print(f"df_LDA сохранён: {len(df_LDA)} точек (по 1000 на класс)")

# Визуализация всех пар признаков
pairs = list(itertools.combinations(range(4), 2))
colors = {0: 'steelblue', 1: 'orange'}

for i, j in pairs:
    plt.figure(figsize=(5.5, 5))
    for c in [0, 1]:
        mask = df_LDA['class'] == c
        # Используем .loc вместо .iloc с булевой маской
        plt.scatter(df_LDA.loc[mask, cols[i]], 
                    df_LDA.loc[mask, cols[j]],
                    c=colors[c], label=f'Класс {c}', alpha=0.7, s=25, edgecolors='k', linewidth=0.3)
    
    plt.xlabel(cols[i])
    plt.ylabel(cols[j])
    plt.title(f"{cols[i]} vs {cols[j]}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df_LDA = pd.read_csv("./df_LDA.csv")

X = df_LDA.drop('class', axis=1).values
y = df_LDA['class'].values
feature_names = df_LDA.columns[:-1]
cols = list(feature_names)

# Стандартизация + PCA
scaler = StandardScaler()
X_std = scaler.fit_transform(X)

pca = PCA()
X_pca = pca.fit_transform(X_std)
explained_var = pca.explained_variance_ratio_

# Цвета по классам
colors = ['steelblue', 'orange']

# Глобальные пределы для единообразия
global_min = X_pca.min() * 1.1
global_max = X_pca.max() * 1.1

# 1. Scree plot
plt.figure(figsize=(6,4))
comps = np.arange(1, 5)
plt.plot(comps, explained_var, 'o-', linewidth=2, color='blue', label='Доля дисперсии')
plt.plot(comps, np.cumsum(explained_var), 's-', linewidth=2, color='red', label='Накопленная доля')
plt.xlabel('Номер главной компоненты')
plt.ylabel('Доля объяснённой дисперсии')
plt.title('Scree plot — df_LDA')
plt.legend()
plt.grid(True)
plt.show()

# Все пары ГК
pc_pairs = [(0,1), (0,2), (0,3), (1,2), (1,3), (2,3)]

# Единый масштаб для стрелок (адаптивный, но одинаковый в пределах датасета)
data_range = np.ptp(X_pca, axis=0).max()
desired_len = data_range * 0.35
max_loading = np.abs(pca.components_).max()
scale_factor = desired_len / max(1.0, max_loading)

# Корреляции переменных с ГК
corr = np.corrcoef(X_std.T, X_pca.T)[:4, 4:]

for i, j in pc_pairs:
    fig, axs = plt.subplots(1, 3, figsize=(16, 5.3))
    fig.suptitle(f"df_LDA: PC{i+1} vs PC{j+1}  ({explained_var[i]:.1%} + {explained_var[j]:.1%})", fontsize=14)

    # 2. Только точки (с раскраской по классам)
    for c in [0, 1]:
        mask = (y == c)
        axs[0].scatter(X_pca[mask, i], X_pca[mask, j],
                       c=colors[c], label=f'Класс {c}', alpha=0.7, s=25, edgecolors='k', linewidth=0.3)
    axs[0].set_title("Точки (по классам)")
    axs[0].legend()
    axs[0].grid(True)

    # 3. Точки + векторы переменных (loadings, одинаковый размер!)
    for c in [0, 1]:
        mask = (y == c)
        axs[1].scatter(X_pca[mask, i], X_pca[mask, j],
                       c=colors[c], alpha=0.7, s=25, edgecolors='k', linewidth=0.3)
    for k in range(4):
        xi = pca.components_[i, k] * scale_factor
        yj = pca.components_[j, k] * scale_factor
        axs[1].arrow(0, 0, xi, yj, color='red', width=0.02, head_width=0.15, head_length=0.15)
        axs[1].text(xi*1.15, yj*1.15, cols[k], color='red', ha='center', fontsize=12)
    axs[1].set_title("Точки + векторы переменных")
    axs[1].grid(True)

    # 4. Круг корреляций
    circle = plt.Circle((0,0), 1, color='gray', fill=False, ls='--', lw=1)
    axs[2].add_artist(circle)
    for k in range(4):
        axs[2].arrow(0, 0, corr[k,i], corr[k,j], color='red', width=0.01,
                    head_width=0.08, head_length=0.08)
        axs[2].text(corr[k,i]*1.15, corr[k,j]*1.15, cols[k], color='red', ha='center', fontsize=12)
    axs[2].set_xlim(-1.2, 1.2)
    axs[2].set_ylim(-1.2, 1.2)
    axs[2].set_aspect('equal')
    axs[2].set_title("Круг корреляций")
    axs[2].grid(True)
    axs[2].axhline(0, color='k', lw=0.8)
    axs[2].axvline(0, color='k', lw=0.8)

    for ax in axs:
        ax.set_xlabel(f"PC{i+1}")
        ax.set_ylabel(f"PC{j+1}")

    plt.tight_layout()
    plt.show()
```

### [7]

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
import itertools

df_LDA = pd.read_csv("./df_LDA.csv")
X = df_LDA.drop('class', axis=1).values
y = df_LDA['class'].values
feature_names = df_LDA.columns[:-1].tolist()

# Обучаем LDA (в 4D → 1D проекция)
lda = LDA(solver='svd')        # svd — самый стабильный для этого случая
X_lda = lda.fit_transform(X, y)   # результат: (2000, 1)

# Параметры для графиков
pairs = list(itertools.combinations(range(4), 2))
colors = ['steelblue', 'orange']
markers = ['o', 's']

# Функция для построения разделяющей прямой на паре признаков
def plot_lda_decision_boundary(ax, i, j):
    # Коэффициенты разделяющей гиперплоскости
    w = lda.coef_[0]                   # вектор весов (4,)
    b = lda.intercept_[0]
    
    # Создаём сетку по двум выбранным признакам
    x_min, x_max = X[:, i].min() - 0.01, X[:, i].max() + 0.01
    y_min, y_max = X[:, j].min() - 0.01, X[:, j].max() + 0.01
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 400),
                         np.linspace(y_min, y_max, 400))
    
    # Заполняем остальные два признака средними значениями (для проекции)
    grid = np.c_[xx.ravel(), yy.ravel()]
    full_grid = np.zeros((grid.shape[0], 4))
    full_grid[:, i] = grid[:, 0]
    full_grid[:, j] = grid[:, 1]
    # Средние по остальным признакам
    for k in range(4):
        if k not in (i, j):
            full_grid[:, k] = X[:, k].mean()
    
    # Предсказание на сетке
    Z = lda.decision_function(full_grid)
    Z = Z.reshape(xx.shape)
    
    # Рисуем контур и разделяющую линию
    ax.contourf(xx, yy, Z, levels=[Z.min(), 0, Z.max()], 
                colors=['steelblue', 'orange'], alpha=0.15)
    ax.contour(xx, yy, Z, levels=[0], colors='red', linewidths=2, linestyles='-')
    
    # Точки
    for cls in [0, 1]:
        mask = (y == cls)
        ax.scatter(X[mask, i], X[mask, j], c=colors[cls], label=f'Класс {cls}',
                   s=30, edgecolors='k', linewidth=0.4, alpha=0.8)

# Визуализация всех 6 пар признаков
plt.figure(figsize=(15, 10))
for idx, (i, j) in enumerate(pairs, 1):
    ax = plt.subplot(2, 3, idx)
    plot_lda_decision_boundary(ax, i, j)
    ax.set_xlabel(feature_names[i])
    ax.set_ylabel(feature_names[j])
    ax.set_title(f'LDA: {feature_names[i]} vs {feature_names[j]}')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.suptitle('LDA — разделяющие границы на всех парах признаков (df_LDA)', fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# Дополнительно: проекция на 1D LDA-ось
plt.figure(figsize=(8, 4))
for cls in [0, 1]:
    mask = (y == cls)
    plt.scatter(X_lda[mask], np.zeros_like(X_lda[mask]) + cls*0.1,
                c=colors[cls], label=f'Класс {cls}', s=40, edgecolors='k')
plt.axvline(lda.intercept_[0] / -lda.coef_[0][0], color='red', linewidth=3,
            label='Решающее правило (порог)')
plt.yticks([0, 0.1], ['Класс 0', 'Класс 1'])
plt.xlabel('LDA-проекция (1D)')
plt.title('1D проекция данных на ось LDA')
plt.legend()
plt.grid(True, axis='x')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

sns.set(style="whitegrid", font_scale=1.3)

# Признаки (784 пикселя, без 'label')
X = df_mnist.drop(columns='label')
feature_names = X.columns.tolist()

# Стандартизация
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Собственные значения и объяснённая дисперсия
eigenvalues = pca.explained_variance_
explained_var = pca.explained_variance_ratio_
cumulative_var = np.cumsum(explained_var)
n_components = len(eigenvalues)

print(f'Количество признаков: {n_components}')
print('Собственные значения:')
for i, val in enumerate(eigenvalues[:20], 1):  # Показываем первые 20 для краткости
    print(f'PC{i:2d}: {val:.4f}')
print('... (и далее)')

# 1. Scree plot + метод ломаной трости + Кайзера
fig, ax = plt.subplots(figsize=(10, 6))
comps = np.arange(1, n_components + 1)
ax.plot(comps, eigenvalues, 'o-', linewidth=2, color='blue', label='Собственные значения')
ax.axhline(y=1, color='red', linestyle='--', label='Критерий Кайзера (λ = 1)')
ax.set_xlabel('Номер главной компоненты')
ax.set_ylabel('Собственное значение')
ax.set_title('Scree plot (Метод ломаной трости + Критерий Кайзера)')
ax.legend()
ax.grid(True)
plt.xlim(0, 50)  # Ограничим для видимости изгиба
plt.show()

# 2. Кумулятивная объяснённая дисперсия
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(comps, cumulative_var * 100, 's-', linewidth=2, color='green')
ax.set_xlabel('Количество главных компонент')
ax.set_ylabel('Накопленная дисперсия (%)')
ax.set_title('Кумулятивная объяснённая дисперсия')
ax.grid(True)

# Отметки 80%, 90%, 95%
for thresh in [80, 90, 95]:
    n = np.argmax(cumulative_var * 100 >= thresh) + 1
    ax.axhline(y=thresh, color='gray', linestyle=':', alpha=0.7)
    ax.axvline(x=n, color='gray', linestyle=':', alpha=0.7)
    print(f"{thresh}% дисперсии объясняется {n} компонентами")
plt.show()

# 3. Выводы
print("\n=== Рекомендации по количеству компонент ===")

# Критерий Кайзера
kaiser = np.sum(eigenvalues > 1)
print(f'1. Критерий Кайзера (λ > 1): {kaiser} компонент')

# Метод ломаной трости (визуально, для MNIST ~40-50)
elbow = 40  # Примерно по изгибу на scree plot (подкорректировать визуально)
print(f'2. Метод ломаной трости: ~{elbow} компонент')

# 90–95%
n90 = np.argmax(cumulative_var >= 0.90) + 1
n95 = np.argmax(cumulative_var >= 0.95) + 1
print(f'3. 90% дисперсии: {n90} компонент')
print(f'   95% дисперсии: {n95} компонент')

print(f'\nДля регрессионной модели: {max(kaiser, elbow, n90)}-{n95} компонент')
```

## [10]

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from itertools import combinations

sns.set(style="whitegrid", font_scale=1.2)

# === Данные ===
X = df_mnist.drop(columns='label')
y = df_mnist['label']

# Стандартизация + PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA()
X_pca = pca.fit_transform(X_scaled)
explained_var = pca.explained_variance_ratio_

# === Определяем, сколько компонент важно ===
n90 = np.argmax(np.cumsum(explained_var) >= 0.90) + 1
n_kaiser = np.sum(pca.explained_variance_ > 1)

print(f"90% дисперсии → {n90} компонент")
print(f"Критерий Кайзера → {n_kaiser} компонент")

# Будем визуализировать только ключевые пары:
# PC1–PC2, PC1–PC3, PC2–PC3, PC3–PC4 — самые информативные
selected_pairs = [(0,1), (0,2), (1,2), (2,3)]
pair_names = ["PC1 vs PC2", "PC1 vs PC3", "PC2 vs PC3", "PC3 vs PC4"]

# Глобальные пределы для точек
global_min = X_pca[:, :4].min() * 1.1
global_max = X_pca[:, :4].max() * 1.1

# Единый адаптивный масштаб для стрелок
max_loading = np.abs(pca.components_[:4, :]).max()
scale_factor = (global_max - global_min) * 0.35 / max(1.0, max_loading)

# Корреляции признаков с PC (для круга корреляций)
corr_with_pc = np.corrcoef(X_scaled.T, X_pca[:, :10].T)[:784, 784:]

print("\nЗапуск визуализаций для 4 ключевых пар главных компонент...")

for idx, (i, j) in enumerate(selected_pairs):
    fig, axs = plt.subplots(1, 3, figsize=(18, 5.5))
    fig.suptitle(f"MNIST — {pair_names[idx]} "
                 f"({explained_var[i]:.1%} + {explained_var[j]:.1%} = {explained_var[i]+explained_var[j]:.1%} дисперсии)",
                 fontsize=16, y=0.98)

    # 1. Только точки (раскраска по цифрам)
    scatter = axs[0].scatter(X_pca[:, i], X_pca[:, j],
                             c=y, cmap='tab10', s=15, alpha=0.8, edgecolor='none')
    axs[0].set_title("Точки (по классам цифр 0–9)")
    axs[0].set_xlabel(f"PC{i+1}")
    axs[0].set_ylabel(f"PC{j+1}")
    axs[0].grid(True, alpha=0.3)

    # 2. Biplot: точки + векторы пикселей (только самые значимые!)
    axs[1].scatter(X_pca[:, i], X_pca[:, j],
                   c=y, cmap='tab10', s=12, alpha=0.7, edgecolor='none')
    
    # Показываем только топ-10 самых сильных loadings для читаемости
    loadings = pca.components_[[i, j], :]
    top_idx = np.argsort(np.abs(loadings).sum(axis=0))[-10:]
    for k in top_idx:
        xi = pca.components_[i, k] * scale_factor
        yj = pca.components_[j, k] * scale_factor
        axs[1].arrow(0, 0, xi, yj, color='red', width=0.02,
                    head_width=0.15, head_length=0.15, alpha=0.9)
        axs[1].text(xi*1.2, yj*1.2, f"p{k}", color='red', fontsize=9, ha='center')
    
    axs[1].set_title("Biplot (точки + топ-пиксели)")
    axs[1].set_xlabel(f"PC{i+1}")
    axs[1].set_ylabel(f"PC{j+1}")
    axs[1].grid(True, alpha=0.3)

    # 3. Круг корреляций
    circle = plt.Circle((0,0), 1, color='gray', fill=False, ls='--', lw=1.5)
    axs[2].add_artist(circle)
    for k in top_idx:
        axs[2].arrow(0, 0, corr_with_pc[k,i], corr_with_pc[k,j],
                    color='red', width=0.01, head_width=0.06)
        axs[2].text(corr_with_pc[k,i]*1.15, corr_with_pc[k,j]*1.15,
                   f"p{k}", color='red', fontsize=9, ha='center')
    axs[2].set_xlim(-1.2, 1.2)
    axs[2].set_ylim(-1.2, 1.2)
    axs[2].set_aspect('equal')
    axs[2].set_title("Круг корреляций (топ-пиксели)")
    axs[2].grid(True, alpha=0.3)
    axs[2].axhline(0, color='k', lw=0.8)
    axs[2].axvline(0, color='k', lw=0.8)

    # Легенда по цифрам — одна на все подграфики
    handles, labels = scatter.legend_elements()
    fig.legend(handles, range(10), title="Цифра", loc="upper right", fontsize=10, ncol=2)

    plt.tight_layout()
    plt.show()

print("Вывод: даже первые 2–3 компоненты отлично разделяют цифры (особенно 0, 1, 4, 7, 9)!")
```

### [11]

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import umap
import pacmap
from sklearn.manifold import TSNE

sns.set(style="whitegrid", font_scale=1.3)

# === Выборка 15 000 точек ===
np.random.seed(42)
idx = np.random.choice(len(df_mnist), 15_000, replace=False)
X_sample = df_mnist.drop('label', axis=1).iloc[idx].values
y_sample = df_mnist['label'].iloc[idx].values

print(f"Работаем с {len(X_sample):,} случайными изображениями из MNIST")

# Стандартизация
X_scaled = StandardScaler().fit_transform(X_sample)

# PCA до 50 компонент
X_pca50 = PCA(n_components=50, random_state=42).fit_transform(X_scaled)
print(f"PCA(50) объясняет {X_pca50.shape[1]} компонент → {PCA(n_components=50).fit(X_scaled).explained_variance_ratio_.sum():.1%} дисперсии")

# === Параметры ===
random_states = [0, 42]

# === Визуализация: 3 строки × 3 колонки ===
fig, axes = plt.subplots(3, 3, figsize=(18, 15))
fig.suptitle('MNIST (15k): Сравнение t-SNE, UMAP, PaCMAP\n(2 random state без PCA + 1 после PCA)', 
             fontsize=18, y=0.96)

methods = [
    ("t-SNE",   lambda rs: TSNE(n_components=2, random_state=rs, perplexity=30, max_iter=1000, init='pca')),
    ("UMAP",    lambda rs: umap.UMAP(n_components=2, random_state=rs, n_neighbors=15, min_dist=0.1)),
    ("PaCMAP",  lambda rs: pacmap.PaCMAP(n_components=2, random_state=rs, n_neighbors=15, MN_ratio=0.5, FP_ratio=2.0))
]

for row, (name, reducer_func) in enumerate(methods):
    # Два запуска без PCA
    for col, rs in enumerate(random_states):
        reducer = reducer_func(rs)
        emb = reducer.fit_transform(X_scaled)
        title = f"{name} (rs={rs})"
        
        scatter = axes[row, col].scatter(emb[:, 0], emb[:, 1], 
                                         c=y_sample, cmap='tab10', s=12, alpha=0.85, linewidth=0)
        axes[row, col].set_title(title, fontsize=14)
        axes[row, col].set_xticks([]); axes[row, col].set_yticks([])

    # Один запуск после PCA
    if name == "t-SNE":
        emb_pca = TSNE(n_components=2, random_state=42, perplexity=30, max_iter=1000).fit_transform(X_pca50)
    elif name == "UMAP":
        emb_pca = umap.UMAP(n_components=2, random_state=42, n_neighbors=15, min_dist=0.1).fit_transform(X_pca50)
    else:  # PaCMAP
        emb_pca = pacmap.PaCMAP(n_components=2, random_state=42, n_neighbors=15).fit_transform(X_pca50)
    
    axes[row, 2].scatter(emb_pca[:, 0], emb_pca[:, 1], 
                         c=y_sample, cmap='tab10', s=12, alpha=0.85, linewidth=0)
    axes[row, 2].set_title(f"{name} после PCA(50)", fontsize=14)
    axes[row, 2].set_xticks([]); axes[row, 2].set_yticks([])

# Единая легенда
handles, labels = scatter.legend_elements()
fig.legend(handles, range(10), title="Цифра", loc="upper right", fontsize=12, markerscale=2, ncol=2)

plt.tight_layout()
plt.show()
```
