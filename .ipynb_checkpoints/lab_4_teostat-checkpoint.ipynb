### Tasl 1: 5-Dim DataFrame Generating

```python
n_samples = 1000
A = 10.0
k = 5

# Среднее направление линии: от (0, 0, 0, 0, 0) до (+A, +A, +A, +A, +A)
t = np.linspace(0, 1, n_samples)
mean_values = np.outer(t, np.ones(5) * A) # матрица n_samples × 5

# Стандартные отклонения для каждой оси 5-мерного пространства:
sigma = np.array([
    np.sqrt(A / k),        # ось 1
    np.sqrt(A / k),        # ось 2
    np.sqrt(A / (k * 2)),  # ось 3
    np.sqrt(A / (k * 4)),  # ось 4
    np.sqrt(A / (k * 8))   # ось 5
])

# Сгенериируем немного шума для каждой оси:
noise = np.random.normal(0, sigma, size=(n_samples, 5))

# Формируем итоговый датасет из сгенерированных данных:
data = mean_values + noise

# Создаём DataFrame для удобства
columns = [f'X{i+1}' for i in range(5)]
df = pd.DataFrame(data, columns=columns)
```

### Task 2: df_out_i DataFrames Generating

Для будущего эксперимента нам нужно будет создать 4 доп. разных версий датасета `df`, где с разной кооридантой лежат
1000 доп. точек того же класса.

```python
# Список значений i:
i_values = [1, 2, 5, 10]

# Словарь для хранения датасетов:
df_out_dict = {}

# Features:
columns = df.columns.tolist()  # ['X1', 'X2', 'X3', 'X4', 'X5']

for i in i_values:
    center = np.array([A*i, -A*i, A*i, -A*i, A*i])
    
    # 1000 абсолютно одинаковых точек:
    outlier_points = np.tile(center, (1000, 1)) # shape (1000, 5)
    
    df_outlier = pd.DataFrame(outlier_points, columns=columns)
    
    # Объединяем основной датасет и новые точки:
    df_combined = pd.concat([df, df_outlier], ignore_index=True)
    df_out_dict[i] = df_combined.copy()
    
    # ----------------- Визуализация -----------------
    print(f'\n=== df_out_{i} (i={i}, 1000 new samples added {center}) ===')
    
    # Создаём маску: True -> основные точки, False -> добавленные:
    is_main = np.ones(len(df_combined), dtype=bool)
    is_main[-1000:] = False
    
    g = sns.pairplot(
        df_combined,
        diag_kind='hist',
        plot_kws={
            'alpha': 0.6,
            'edgecolor': 'none'
        },
        diag_kws={
            'bins': 40,
            'alpha': 0.7,
            'edgecolor': 'black',
            'linewidth': 0.5,
            'kde': True,
            'line_kws': {'linewidth': 2.0, 'color': '#2ca02c'}
        },
        height=2.8,
        aspect=1
    )
    
    for ax in g.axes.flat:
        if ax is None:
            continue
        
        x = ax.get_xlabel()
        y = ax.get_ylabel()
        
        if x == "" or y == "":
            continue
        
        # Основные точки:
        ax.scatter(
            df_combined.loc[is_main, x],
            df_combined.loc[is_main, y],
            c='#1f77b4', alpha=0.6, s=20, edgecolor='none'
        )
        
        # Добавленные 1000 идентичных точек:
        ax.scatter(
            df_combined.loc[~is_main, x],
            df_combined.loc[~is_main, y],
            c='#ff7f0e', alpha=1.0, s=100,
            edgecolor='black', linewidth=1,
            marker='o', label='Added samples'
        )

    
    # Добавляем легенду:
    handles, labels = g.axes[-1,-1].get_legend_handles_labels()
    g.fig.legend(handles, labels, loc='upper right')
    
    g.fig.suptitle(
        f'df_out_{i} Dataset (A={A}, k={k}, i={i})\n'
        f'Base samples (blue) + 1000 the same ones in (+{A*i}, -{A*i}, +{A*i}, -{A*i}, +{A*i}) (orange)',
        fontsize=15, y=0.98
    )
    plt.subplots_adjust(top=0.92)
    plt.show()
```

### Task 3: PCA Processing

```python
# Список всех датасетов для обработки:
datasets = {
    'df_base': df,
    'df_out_1': df_out_dict[1],
    'df_out_2': df_out_dict[2],
    'df_out_5': df_out_dict[5],
    'df_out_10': df_out_dict[10]
}
```

```python
sns.set(style="whitegrid", font_scale=1.1)

all_pairs = list(combs(range(5), 2))
for name, data in datasets.items():
    print(f'\n{"="*60}')
    print(f'PROCESSING: {name}')
    print(f'{"="*60}')
    
    X = data.copy()
    feature_names = X.columns.tolist()
    
    # Маска основных и добавленных точек:
    if 'out' in name.lower():
        n_main = len(df)
        is_main = np.ones(len(X), dtype=bool)
        is_main[n_main:] = False
    else:
        is_main = np.ones(len(X), dtype=bool)
    
    # Стандартизация и PCA:
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    pca = PCA()
    X_pca = pca.fit_transform(X_scaled)
    
    explained_var = pca.explained_variance_ratio_
    
    # Глобальные пределы для проекций и biplot:
    global_min = X_pca.min() * 1.1
    global_max = X_pca.max() * 1.1
    
    # ----------------- 1. Scree plot -----------------
    fig, ax = plt.subplots(1, 1, figsize=(8, 5))
    comps = np.arange(1, len(explained_var) + 1)
    ax.plot(comps, explained_var, 'o-', linewidth=2, color='blue', label='The proportion of variance')
    ax.plot(comps, np.cumsum(explained_var), 's-', linewidth=2, color='red', label='Accumulated share')
    ax.set_xlabel('Number of the main component')
    ax.set_ylabel('The proportion of the explained variance')
    ax.set_title(f'Scree plot - {name}')
    ax.legend()
    ax.grid(True)
    plt.show()
    
    # ----------------- 2. Все пары ГК — только точки -----------------
    fig, axes = plt.subplots(2, 5, figsize=(25, 10))
    axes = axes.flatten()
    
    for idx, (i, j) in enumerate(all_pairs):
        ax = axes[idx]
        
        ax.scatter(X_pca[is_main, i], X_pca[is_main, j],
                   c='#1f77b4', alpha=0.7, s=20, edgecolor='none', label='Base samples')
        
        if not np.all(is_main):
            ax.scatter(X_pca[~is_main, i], X_pca[~is_main, j],
                       c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2,
                       label='1000 added samples')
        
        ax.set_xlabel(f'PC{i+1} ({explained_var[i]:.1%})')
        ax.set_ylabel(f'PC{j+1} ({explained_var[j]:.1%})')
        ax.set_title(f'PC{i+1} vs PC{j+1}')
        ax.set_xlim(global_min, global_max)
        ax.set_ylim(global_min, global_max)
        ax.legend(fontsize=9)
    
    plt.suptitle(f'All pairs of principal components (points only) - {name}', fontsize=18, y=0.98)
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
    
    # ----------------- 3. Все пары ГК — biplot (с адаптивным масштабом стрелок) -----------------
    fig, axes = plt.subplots(2, 5, figsize=(25, 10))
    axes = axes.flatten()
    
    for idx, (i, j) in enumerate(all_pairs):
        ax = axes[idx]
        
        # Samples drawing:
        ax.scatter(X_pca[is_main, i], X_pca[is_main, j],
                   c='#1f77b4', alpha=0.5, s=15, label='Base samples')
        if not np.all(is_main):
            ax.scatter(X_pca[~is_main, i], X_pca[~is_main, j],
                       c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2,
                       label='1000 added samples')
        
        # Адаптивный масштаб для стрелок:
        max_loading = np.max(np.abs(pca.components_[[i, j], :]))
        if max_loading == 0:
            max_loading = 1
        desired_max_length = abs(global_max - global_min) * 0.35
        scale_factor = desired_max_length / max_loading
        
        # Векторы переменных с адаптивным масштабом:
        for k in range(pca.components_.shape[1]):
            xi = pca.components_[i, k] * scale_factor
            yj = pca.components_[j, k] * scale_factor
            ax.arrow(0, 0, xi, yj,
                     color='black', alpha=0.8, width=0.02,
                     head_width=0.15, head_length=0.15)
            ax.text(xi * 1.15, yj * 1.15,
                    feature_names[k], color='black', ha='center', va='center', fontsize=11)
        
        ax.set_xlabel(f'PC{i+1} ({explained_var[i]:.1%})')
        ax.set_ylabel(f'PC{j+1} ({explained_var[j]:.1%})')
        ax.set_title(f'Biplot PC{i+1} vs PC{j+1}')
        ax.set_xlim(global_min, global_max)
        ax.set_ylim(global_min, global_max)
        ax.legend(fontsize=9)
    
    plt.suptitle(f'All pairs of main components (biplot) - {name}', fontsize=18, y=0.98)
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
    
    # ----------------- 4. Круг корреляций — все 10 пар -----------------
    loadings = pca.components_.T
    
    fig, axes = plt.subplots(2, 5, figsize=(25, 10))
    axes = axes.flatten()
    
    for idx, (i, j) in enumerate(all_pairs):
        ax = axes[idx]
        
        # Единичная окружность:
        circle = plt.Circle((0, 0), 1, color='gray', fill=False, linestyle='--')
        ax.add_artist(circle)
        
        # Стрелки корреляций:
        for k in range(loadings.shape[0]):
            ax.arrow(0, 0, loadings[k, i], loadings[k, j],
                     color='red', alpha=0.8, width=0.015)
            ax.text(loadings[k, i] * 1.15, loadings[k, j] * 1.15,
                    feature_names[k], color='red', ha='center', fontsize=11)
        
        ax.set_xlim(-1.2, 1.2)
        ax.set_ylim(-1.2, 1.2)
        ax.set_aspect('equal')
        ax.set_xlabel(f'PC{i+1}')
        ax.set_ylabel(f'PC{j+1}')
        ax.set_title(f'Correlations with PC{i+1} and PC{j+1}')
        ax.grid(True)
    
    plt.suptitle(f'Circle of correlations (all pairs) -> {name}', fontsize=18, y=0.98)
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
```

### Task 4: UMAP, PacMAP, tSNE 

```python
sns.set(style="whitegrid", font_scale=1.2)

# Количество случайных запусков:
n_repeats = 4
random_states = [0, 42, 404, 2112005]

# Предварительное PCA (до 5 компонент, т.к. у нас всего 5 признаков):
pca_pre = PCA(n_components=3)

for name, data in datasets.items():
    print(f'\n{"="*70}')
    print(f'{name} processing...')
    print(f'{"="*70}')
    
    X = data.copy().values
    
    # Маска добавленных точек:
    if 'out' in name.lower():
        n_main = len(df)
        is_main = np.zeros(len(X), dtype=bool)
        is_main[:n_main] = True
    else:
        is_main = np.ones(len(X), dtype=bool)
    # Один раз на датасет: стандартизация + предварительное PCA:
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(data)
    X_pca_pre = pca_pre.fit_transform(X_scaled)
    
    # По 4 случайных запуска + после PCA:
    # ----------------- 1. t-SNE -----------------
    fig, axes = plt.subplots(1, 5, figsize=(25, 5))
    for idx, rs in enumerate(random_states):
        tsne = TSNE(n_components=2, random_state=rs, perplexity=30)
        X_emb = tsne.fit_transform(X)
      
        ax = axes[idx]
        ax.scatter(X_emb[is_main, 0], X_emb[is_main, 1],
                   c='#1f77b4', alpha=0.7, s=20, label='Base samples')
        if not np.all(is_main):
            ax.scatter(X_emb[~is_main, 0], X_emb[~is_main, 1],
                       c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2,
                       label='1000 added samples')
        ax.set_title(f't-SNE (rs={rs})')
    
    # t-SNE после PCA:
    tsne_pca = TSNE(n_components=2, random_state=42, perplexity=30)
    X_emb_pca = tsne_pca.fit_transform(X_pca_pre)
    ax = axes[4]
    ax.scatter(X_emb_pca[is_main, 0], X_emb_pca[is_main, 1],
               c='#1f77b4', alpha=0.7, s=20)
    if not np.all(is_main):
        ax.scatter(X_emb_pca[~is_main, 0], X_emb_pca[~is_main, 1],
                   c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2)
    ax.set_title('t-SNE after PCA')
    
    # Общая легенда:
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc='upper right', fontsize=12)
    
    plt.suptitle(f't-SNE - {name}', fontsize=16)
    plt.tight_layout()
    plt.show()
    
    # ----------------- 2. UMAP -----------------
    fig, axes = plt.subplots(1, 5, figsize=(25, 5))
    for idx, rs in enumerate(random_states):
        reducer = umap.UMAP(n_components=2, random_state=rs, n_neighbors=15, min_dist=0.1)
        X_emb = reducer.fit_transform(X)
      
        ax = axes[idx]
        ax.scatter(X_emb[is_main, 0], X_emb[is_main, 1],
                   c='#1f77b4', alpha=0.7, s=20)
        if not np.all(is_main):
            ax.scatter(X_emb[~is_main, 0], X_emb[~is_main, 1],
                       c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2)
        ax.set_title(f'UMAP (rs={rs})')
    
    # UMAP после PCA:
    reducer_pca = umap.UMAP(n_components=2, random_state=42, n_neighbors=15, min_dist=0.1)
    X_emb_pca = reducer_pca.fit_transform(X_pca_pre)
    ax = axes[4]
    ax.scatter(X_emb_pca[is_main, 0], X_emb_pca[is_main, 1],
               c='#1f77b4', alpha=0.7, s=20)
    if not np.all(is_main):
        ax.scatter(X_emb_pca[~is_main, 0], X_emb_pca[~is_main, 1],
                   c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2)
    ax.set_title('UMAP after PCA')
    
    # Общая легенда:
    if not np.all(is_main):
        handles, labels = axes[0].get_legend_handles_labels()
        fig.legend(handles, labels, loc='upper right', fontsize=12)
    
    plt.suptitle(f'UMAP - {name}', fontsize=16)
    plt.tight_layout()
    plt.show()
    
    # ----------------- 3. PaCMAP -----------------
    fig, axes = plt.subplots(1, 5, figsize=(25, 5))
    for idx, rs in enumerate(random_states):
        reducer = pacmap.PaCMAP(n_components=2, random_state=rs, n_neighbors=10, MN_ratio=0.5, FP_ratio=2.0)
        X_emb = reducer.fit_transform(X)
      
        ax = axes[idx]
        ax.scatter(X_emb[is_main, 0], X_emb[is_main, 1],
                   c='#1f77b4', alpha=0.7, s=20)
        if not np.all(is_main):
            ax.scatter(X_emb[~is_main, 0], X_emb[~is_main, 1],
                       c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2)
        ax.set_title(f'PaCMAP (rs={rs})')
    
    # PaCMAP после PCA:
    reducer_pca = pacmap.PaCMAP(n_components=2, random_state=42, n_neighbors=10, MN_ratio=0.5, FP_ratio=2.0)
    X_emb_pca = reducer_pca.fit_transform(X_pca_pre)
    ax = axes[4]
    ax.scatter(X_emb_pca[is_main, 0], X_emb_pca[is_main, 1],
               c='#1f77b4', alpha=0.7, s=20)
    if not np.all(is_main):
        ax.scatter(X_emb_pca[~is_main, 0], X_emb_pca[~is_main, 1],
                   c='#ff7f0e', alpha=1.0, s=120, edgecolor='black', linewidth=1.2)
    ax.set_title('PaCMAP after PCA')
    
    # Общая легенда:
    if not np.all(is_main):
        handles, labels = axes[0].get_legend_handles_labels()
        fig.legend(handles, labels, loc='upper right', fontsize=12)
    
    plt.suptitle(f'PaCMAP - {name}', fontsize=16)
    plt.tight_layout()
    plt.show()
```

### Task 5: df_LDA DataFrame Generating

```python
# Параметры:
A = 10.0
n_samples_per_class = 1000
n_features = 4

# Дисперсия шума для каждого класса (одинаковая по всем осям):
sigma_squared = A / 200000
sigma = np.sqrt(sigma_squared) # std

# Class 0: линия от (0,0,0,0) до (A,A,A,A):
t0 = np.linspace(0, 1, n_samples_per_class)
mean0 = np.outer(t0, np.ones(n_features) * A) # shape (1000, 4)

noise0 = np.random.normal(0, sigma, size=(n_samples_per_class, n_features))
class0 = mean0 + noise0
labels0 = np.zeros(n_samples_per_class)

# Class 1: линия от (A/100, 0, 0, 0) до (A + A/100, A, A, A)
start1 = np.array([A/100, 0, 0, 0])
end1   = np.array([A + A/100, A, A, A])
direction1 = end1 - start1

t1 = np.linspace(0, 1, n_samples_per_class)
mean1 = np.outer(t1, direction1) + start1 # shape (1000, 4)

noise1 = np.random.normal(0, sigma, size=(n_samples_per_class, n_features))
class1 = mean1 + noise1
labels1 = np.ones(n_samples_per_class)

# Объединяем:
X_lda = np.vstack([class0, class1])
y_lda = np.hstack([labels0, labels1])

columns = [f'X{i+1}' for i in range(n_features)]
df_LDA = pd.DataFrame(X_lda, columns=columns)
df_LDA['class'] = y_lda.astype(int)

print('df_LDA was created!')
```

### Task 6: df_LDA + PCA Processing

```python
sns.set(style="whitegrid", font_scale=1.1)

# Датасет только `df_LDA` (4 признака + класс):
data = df_LDA.copy()
name = 'df_LDA'
print(f'\n{"="*60}')
print(f'PROCESSING: {name}')
print(f'{"="*60}')

X = data.drop(columns='class')
y = data['class']
feature_names = X.columns.tolist()

# Стандартизация и PCA:
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

explained_var = pca.explained_variance_ratio_

# Глобальные пределы для одинакового масштаба:
global_min = X_pca.min() * 1.1
global_max = X_pca.max() * 1.1

# Все пары компонент (для 4 признаков - 6 пар):
all_pairs = list(combs(range(4), 2))

# ----------------- 1. Scree plot -----------------
fig, ax = plt.subplots(1, 1, figsize=(8, 5))
comps = np.arange(1, len(explained_var) + 1)
ax.plot(comps, explained_var, 'o-', linewidth=2, color='blue', label='The proportion of variance')
ax.plot(comps, np.cumsum(explained_var), 's-', linewidth=2, color='red', label='Accumulated share')
ax.set_xlabel('Number of the main component')
ax.set_ylabel('The proportion of the explained variance')
ax.set_title(f'Scree plot - {name}')
ax.legend()
ax.grid(True)
plt.show()

# ----------------- 2. Все пары ГК — только точки (с окраской по классам) -----------------
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for idx, (i, j) in enumerate(all_pairs):
    ax = axes[idx]
    scatter = ax.scatter(X_pca[:, i], X_pca[:, j],
                         c=y, cmap='coolwarm', alpha=0.7, s=30, edgecolor='none')
    ax.set_xlabel(f'PC{i+1} ({explained_var[i]:.1%})')
    ax.set_ylabel(f'PC{j+1} ({explained_var[j]:.1%})')
    ax.set_title(f'PC{i+1} vs PC{j+1}')
    ax.set_xlim(global_min, global_max)
    ax.set_ylim(global_min, global_max)

# Легенда по классам:
handles, labels = scatter.legend_elements()
ax.legend(handles, ['Class 0', 'Class 1'], loc='upper right', fontsize=10)

plt.suptitle(f'All pairs of principal components (points only) - {name}', fontsize=18, y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# ----------------- 3. Все пары ГК — biplot (точки + векторы, с адаптивным масштабом) -----------------
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for idx, (i, j) in enumerate(all_pairs):
    ax = axes[idx]
    
    # Точки с окраской по классам:
    scatter = ax.scatter(X_pca[:, i], X_pca[:, j],
                         c=y, cmap='coolwarm', alpha=0.6, s=25, edgecolor='none')
    
    # Адаптивный масштаб для стрелок:
    max_loading = np.max(np.abs(pca.components_[[i, j], :]))
    if max_loading == 0:
        max_loading = 1
    desired_max_length = abs(global_max - global_min) * 0.35
    scale_factor = desired_max_length / max_loading
    
    # Векторы переменных:
    for k in range(pca.components_.shape[1]):
        xi = pca.components_[i, k] * scale_factor
        yj = pca.components_[j, k] * scale_factor
        ax.arrow(0, 0, xi, yj,
                 color='black', alpha=0.8, width=0.02,
                 head_width=0.15, head_length=0.15)
        ax.text(xi * 1.15, yj * 1.15,
                feature_names[k], color='black', ha='center', va='center', fontsize=11)
    
    ax.set_xlabel(f'PC{i+1} ({explained_var[i]:.1%})')
    ax.set_ylabel(f'PC{j+1} ({explained_var[j]:.1%})')
    ax.set_title(f'Biplot PC{i+1} vs PC{j+1}')
    ax.set_xlim(global_min, global_max)
    ax.set_ylim(global_min, global_max)

# Легенда по классам (одна на весь figure):
handles, labels = scatter.legend_elements()
fig.legend(handles, ['Class 0', 'Class 1'], loc='upper right', fontsize=12)

plt.suptitle(f'All pairs of main components (biplot) - {name}', fontsize=18, y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# ----------------- 4. Круг корреляций — все 6 пар -----------------
loadings = pca.components_.T

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for idx, (i, j) in enumerate(all_pairs):
    ax = axes[idx]
    
    circle = plt.Circle((0, 0), 1, color='gray', fill=False, linestyle='--')
    ax.add_artist(circle)
    
    for k in range(loadings.shape[0]):
        ax.arrow(0, 0, loadings[k, i], loadings[k, j],
                 color='red', alpha=0.8, width=0.015)
        ax.text(loadings[k, i] * 1.15, loadings[k, j] * 1.15,
                feature_names[k], color='red', ha='center', fontsize=11)
    
    ax.set_xlim(-1.2, 1.2)
    ax.set_ylim(-1.2, 1.2)
    ax.set_aspect('equal')
    ax.set_xlabel(f'PC{i+1}')
    ax.set_ylabel(f'PC{j+1}')
    ax.set_title(f'Correlations with PC{i+1} and PC{j+1}')
    ax.grid(True)

plt.suptitle(f'Circle of correlations (all the pairs) - {name}', fontsize=18, y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

### Task 7: LDA IS BACK

```python
sns.set(style="whitegrid", font_scale=1.2)

# Датасет `df_LDA`:
X = df_LDA.drop(columns='class')
y = df_LDA['class']

feature_names = X.columns.tolist()

# Обучаем LDA:
lda = LinearDiscriminantAnalysis()
lda.fit(X, y)

# Коэффициенты разделяющей гиперплоскости:
coef = lda.coef_[0] # *the vector of the normal to the hyperplane*
intercept = lda.intercept_[0]

print(f'LDA coefficients (weights of variables): {coef}')
print(f'Free member: {intercept:.4f}')

# Все пары признаков:
pairs = list(combs(range(4), 2))

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, (i, j) in enumerate(pairs):
    ax = axes[idx]
    
    # Точки с окраской по классам:
    scatter = ax.scatter(X.iloc[:, i], X.iloc[:, j],
                         c=y, cmap='coolwarm', alpha=0.7, s=40, edgecolor='k', linewidth=0.3)
    
    # Сетка для предсказания границы:
    x_min, x_max = X.iloc[:, i].min() - 1, X.iloc[:, i].max() + 1
    y_min, y_max = X.iloc[:, j].min() - 1, X.iloc[:, j].max() + 1
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 400),
                         np.linspace(y_min, y_max, 400))
    
    grid = np.zeros((xx.ravel().shape[0], 4))
    grid[:, i] = xx.ravel()
    grid[:, j] = yy.ravel()
    for k in range(4):
        if k not in (i, j):
            grid[:, k] = X.iloc[:, k].mean()
    
    # Предсказание LDA:
    Z = lda.decision_function(grid)
    Z = Z.reshape(xx.shape)
    
    # Рисуем контур разделяющей гиперплоскости (Z = 0):
    ax.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
    ax.contourf(xx, yy, Z, levels=[Z.min(), 0, Z.max()], alpha=0.15, colors=['#1f77b4', '#ff7f0e'])
    
    ax.set_xlabel(feature_names[i])
    ax.set_ylabel(feature_names[j])
    ax.set_title(f'LDA on the pair {feature_names[i]} vs {feature_names[j]}')

# Легенда по классам:
handles, labels = scatter.legend_elements()
fig.legend(handles, ['Class 0', 'Class 1'], loc='upper right', fontsize=14)

plt.suptitle('LDA solution on all pairs of initial variables (df_LDA)', fontsize=18, y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# Дополнительно: проекция на единственную дискриминантную ось LDA:
X_lda_1d = lda.transform(X) # shape (2000, 1)

fig, ax = plt.subplots(figsize=(10, 6))
for cls in [0, 1]:
    ax.hist(X_lda_1d[y == cls], bins=50, alpha=0.7, label=f'Class {cls}', density=True)
ax.set_xlabel('Projection onto the discriminant LDA axis')
ax.set_ylabel('Density')
ax.set_title('Class distributions on the discriminant LDA axis')
ax.legend()
plt.show()

print(f'\nLDA\'s coefficients: {dict(zip(feature_names, coef))}')
```

### Task 9: The Kaiser Method/The Broken Cane Method

```python
sns.set(style="whitegrid", font_scale=1.3)

# Отделяем признаки (все кроме 'Class'):
X = df_beans.drop(columns='Class')
feature_names = X.columns.tolist()

# Стандартизация:
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA:
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Собственные значения и объяснённая дисперсия:
eigenvalues = pca.explained_variance_
explained_var = pca.explained_variance_ratio_
cumulative_var = np.cumsum(explained_var)

n_components = len(eigenvalues)

print(f'Common number of features: {n_components}')
print('Eigenvalues:')
for i, val in enumerate(eigenvalues, 1):
    print(f'PC{i:2d}: {val:.4f}')

# ----------------- 1. Scree plot + метод ломаной трости -----------------
fig, ax = plt.subplots(figsize=(10, 6))
comps = np.arange(1, n_components + 1)

ax.plot(comps, eigenvalues, 'o-', linewidth=2, color='blue', label='Eigenvalues')
ax.axhline(y=1, color='red', linestyle='--', label='The Kaiser Criterion (λ = 1)')
ax.set_xlabel('Number of the Main component')
ax.set_ylabel('Eigenvalue')
ax.set_title('Scree plot (The Broken Cane Method + the Kaiser Criterion)')
ax.legend()
ax.grid(True)

plt.show()

# ----------------- 2. Кумулятивная объяснённая дисперсия -----------------
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(comps, cumulative_var * 100, 's-', linewidth=2, color='green')
ax.set_xlabel('Number of Main components')
ax.set_ylabel('Accumulated explained variance (%)')
ax.set_title('Cumulative explained variance')
ax.grid(True)

# Отметки 80%, 90%, 95%:
for thresh in [80, 90, 95]:
    n = np.argmax(cumulative_var * 100 >= thresh) + 1
    ax.axhline(y=thresh, color='gray', linestyle=':', alpha=0.7)
    ax.axvline(x=n, color='gray', linestyle=':', alpha=0.7)
    print(f"{thresh}% of variance is explained by {n} components")

plt.show()

# ----------------- 3. Выводы по методам -----------------
print("\n=== Recommendations on the number of components ===")

# Критерий Кайзера:
kaiser = np.sum(eigenvalues > 1)
print(f'1. The Kaiser Criterion (λ > 1): leave {kaiser} component(s)')

# Метод ломаной трости (визуально смотрим на scree plot):
elbow = 5
print(f'2. The Broken Cane Method (kink on the scree plot): ~{elbow} component(s)')

# Практическое правило (90–95%):
n90 = np.argmax(cumulative_var >= 0.90) + 1
n95 = np.argmax(cumulative_var >= 0.95) + 1
print(f'3. 90% of variance: {n90} component(s)')
print(f'   95% of variance: {n95} component(s)')

print(f'\nRecommendation for a regression model: {max(kaiser, elbow, n90)}-{n95} components')
```

### Task 11: df_beans + UMAP, PacMAP, tSNE

```python
sns.set(style="whitegrid", font_scale=1.2)

# Датасет с бобами:
X = df_beans.drop(columns='Class')
y = df_beans['Class']
class_names = y.unique()

# Стандартизация:
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Предварительное PCA до 5 компонент:
pca_pre = PCA(n_components=5)
X_pca_pre = pca_pre.fit_transform(X_scaled)

# Параметры для воспроизводимости:
random_states = [0, 42, 202]

print(f'\n{"="*80}')
print('t-SNE, UMAP, PaCMAP with Dry Bean Dataset (7 classes)')
print(f'{"="*80}')

# ----------------- 1. t-SNE -----------------
fig, axes = plt.subplots(1, 4, figsize=(24, 6))
for idx, rs in enumerate(random_states):
    tsne = TSNE(n_components=2, random_state=rs, perplexity=30, max_iter=1000)
    X_emb = tsne.fit_transform(X_scaled)
    
    ax = axes[idx]
    scatter = ax.scatter(X_emb[:, 0], X_emb[:, 1],
                         c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
    ax.set_title(f't-SNE (rs={rs})')

# t-SNE после PCA:
tsne_pca = TSNE(n_components=2, random_state=42, perplexity=30, max_iter=1000)
X_emb_pca = tsne_pca.fit_transform(X_pca_pre)
ax = axes[3]
ax.scatter(X_emb_pca[:, 0], X_emb_pca[:, 1],
           c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
ax.set_title('t-SNE after PCA (5 comps.)')

plt.suptitle('t-SNE - Dry Bean Dataset', fontsize=16)
handles, labels = scatter.legend_elements()
fig.legend(handles, class_names, loc='upper right', title='Sort of bean', fontsize=10)
plt.tight_layout()
plt.show()

# ----------------- 2. UMAP -----------------
fig, axes = plt.subplots(1, 4, figsize=(24, 6))
for idx, rs in enumerate(random_states):
    reducer = umap.UMAP(n_components=2, random_state=rs, n_neighbors=10, min_dist=0.1)
    X_emb = reducer.fit_transform(X_scaled)
    
    ax = axes[idx]
    scatter = ax.scatter(X_emb[:, 0], X_emb[:, 1],
                         c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
    ax.set_title(f'UMAP (rs={rs})')

# UMAP после PCA:
reducer_pca = umap.UMAP(n_components=2, random_state=42, n_neighbors=10, min_dist=0.1)
X_emb_pca = reducer_pca.fit_transform(X_pca_pre)
ax = axes[3]
ax.scatter(X_emb_pca[:, 0], X_emb_pca[:, 1],
           c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
ax.set_title('UMAP after PCA (5 comps.)')

plt.suptitle('UMAP (n_neighbors=50) - Dry Bean Dataset', fontsize=16)
handles, labels = scatter.legend_elements()
fig.legend(handles, class_names, loc='upper right', title='Sort of bean', fontsize=10)
plt.tight_layout()
plt.show()

# ----------------- 3. PaCMAP -----------------
fig, axes = plt.subplots(1, 4, figsize=(24, 6))
for idx, rs in enumerate(random_states):
    reducer = pacmap.PaCMAP(n_components=2, random_state=rs, n_neighbors=15, 
                            MN_ratio=0.5, FP_ratio=2.0)
    X_emb = reducer.fit_transform(X_scaled)
    
    ax = axes[idx]
    scatter = ax.scatter(X_emb[:, 0], X_emb[:, 1],
                         c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
    ax.set_title(f'PaCMAP (rs={rs})')

# PaCMAP after PCA:
reducer_pca = pacmap.PaCMAP(n_components=2, random_state=42, n_neighbors=15,
                           MN_ratio=0.5, FP_ratio=2.0)
X_emb_pca = reducer_pca.fit_transform(X_pca_pre)
ax = axes[3]
ax.scatter(X_emb_pca[:, 0], X_emb_pca[:, 1],
           c=y.factorize()[0], cmap='tab10', alpha=0.8, s=30)
ax.set_title('PaCMAP after PCA (5 comps.)')

plt.suptitle('PaCMAP (n_neighbors=50) - Dry Bean Dataset', fontsize=16)
handles, labels = scatter.legend_elements()
fig.legend(handles, class_names, loc='upper right', title='Sort of bean', fontsize=10)
plt.tight_layout()
plt.show()
```
